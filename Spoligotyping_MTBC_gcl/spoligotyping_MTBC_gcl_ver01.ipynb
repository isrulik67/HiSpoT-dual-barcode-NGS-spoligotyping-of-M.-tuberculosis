## Locate and Decompress GZ files

### Subtask:
Recursively scan the project folder (`{path}`) and its subdirectories (including those created from `.zip` extraction) to find all `.gz` files. For each `.gz` file, decompress it and save the resulting file with a `.fastq` extension directly into the project folder (`{path}`). The original directory structure of the `.gz` file will not be preserved for the `.fastq` output.


This step of file extraction can take about 2 to 3 minutes.

```python
import os
import gzip
import zipfile
import glob
import time

print(f"Starting file processing in project folder: {path}")

# --- Step 1: Extract any .zip files --- #
print(f"\n--- Searching for and extracting .zip files in '{path}' ---")
zip_files_found = glob.glob(os.path.join(path, '*.zip'))

if not zip_files_found:
    print(f"No .zip files found in '{path}' to extract.")
else:
    print(f"Found {len(zip_files_found)} .zip file(s). Extracting...")
    for zip_file_path in zip_files_found:
        try:
            with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
                print(f"Extracting contents of '{os.path.basename(zip_file_path)}'...")
                zip_ref.extractall(path)
            print(f"Successfully extracted '{os.path.basename(zip_file_path)}' to '{path}'")
        except zipfile.BadZipFile:
            print(f"Error: '{os.path.basename(zip_file_path)}' is not a valid zip file or is corrupted. Skipping.")
        except Exception as e:
            print(f"Error extracting '{os.path.basename(zip_file_path)}': {e}. Skipping.")
    print(".zip extraction process completed.")

# --- Step 2: Recursively find and decompress .gz files --- #
print(f"\n--- Starting recursive decompression of .gz files to .fastq in '{path}' ---")
start_time = time.time()
decompressed_fastq_files = []
gz_files_to_process = []

# First, collect all .gz files to get a total count for progress estimation
for root, _, files in os.walk(path):
    for filename in files:
        if filename.endswith('.gz'):
            gz_files_to_process.append(os.path.join(root, filename))

total_gz_files = len(gz_files_to_process)
print(f"Found {total_gz_files} .gz files to decompress.")

if total_gz_files == 0:
    print("No .gz files found for decompression.")
else:
    for i, gz_file_path in enumerate(gz_files_to_process):
        current_file_num = i + 1
        # Determine the output filename and ensure it ends with .fastq
        output_base_name = os.path.basename(gz_file_path)[:-3] # Remove '.gz'
        if not output_base_name.lower().endswith('.fastq'):
            output_base_name += '.fastq'

        # Ensure output is directly in the main project folder
        output_file_path = os.path.join(path, output_base_name)

        print(f"Processing file {current_file_num} of {total_gz_files}: Decompressing '{os.path.basename(gz_file_path)}' to '{os.path.basename(output_file_path)}'")

        try:
            with gzip.open(gz_file_path, 'rb') as f_in:
                with open(output_file_path, 'wb') as f_out:
                    f_out.write(f_in.read())
            decompressed_fastq_files.append(output_file_path)
        except Exception as e:
            print(f"Error decompressing '{os.path.basename(gz_file_path)}': {e}. Skipping this file.")

    end_time = time.time()
    elapsed_time = end_time - start_time

    print(f"\nAll .gz files have been decompressed, the fastq files are ready for further processing.")
    print(f"Total time elapsed for .gz decompression: {elapsed_time:.2f} seconds.")

    if decompressed_fastq_files:
        print("\nSuccessfully created .fastq files:")
        for fq_file in decompressed_fastq_files:
            print(f"- {os.path.basename(fq_file)}")
    else:
        print("\nNo .gz files were successfully decompressed into .fastq files.")
```

```python
import os

# Get all .fastq files in the project directory
fastq_files = [f for f in os.listdir(path) if f.endswith('.fastq')]

# Separate R1 and R2 files
r1_files = sorted([f for f in fastq_files if '_R1' in f])
r2_files = sorted([f for f in fastq_files if '_R2' in f])

# Combine R1 and R2 files in the specified order
ordered_fastq_files = r1_files + r2_files

# Define the output file path
output_list_file = os.path.join(path, "list_of_file_names.txt")

# Write the list to the text file
with open(output_list_file, "w") as f:
    for filename in ordered_fastq_files:
        f.write(filename + "\n")

print(f"Successfully created '{os.path.basename(output_list_file)}' in '{path}' with the following files (R1 first, then R2):")
for filename in ordered_fastq_files:
    print(f"- {filename}")
```
